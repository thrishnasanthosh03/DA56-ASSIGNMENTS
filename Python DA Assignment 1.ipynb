Numpy Array Operations:   


```python
#1. Create a 1D NumPy Array: 

import numpy as np
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print (temperatures_w1)
```

```python
# 2. Inspection and Properties:

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print (temperatures_w1.shape)
print(temperatures_w1.size)
print(temperatures_w1.dtype)
```

```python
# 3. Array Operations:
# ● Find the maximum, minimum, and mean temperatures for the week.

temperatures_w1_f = (temperatures_w1 * 9/5) + 32
print(temperatures_w1_f)

print(np.max(temperatures_w1_f))
print(np.min(temperatures_w1_f))
print(np.mean(temperatures_w1_f))
```

```python
# 4. Array Slicing and Indexing:
print(temperatures_w1[0:3])
print(temperatures_w1[-2:])
print(temperatures_w1[2:5])
```

```python
# 5. Create a 2D Array:

temperatures=np.array([[22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],[19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]])
print(temperatures)
```

```python
# 6. Inspect and Slice the 2D Array: 

print (temperatures.shape)
print(temperatures.size)
print(temperatures.dtype)

```

```python
# Extract the temperatures for each week and the weekends (last two days) for both weeks.
week_1 = temperatures[0, :]
week_2 = temperatures[1, :]
weekends = temperatures[:, -2:]

print(week_1)
print(week_2)
print(weekends)
```
