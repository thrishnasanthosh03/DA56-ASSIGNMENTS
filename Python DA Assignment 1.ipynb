Numpy Array Operations:   


```python
#1. Create a 1D NumPy Array: 

import numpy as np
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print (temperatures_w1)
```

```python
# 2. Inspection and Properties:

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print (temperatures_w1.shape)
print(temperatures_w1.size)
print(temperatures_w1.dtype)
```

```python
# 3. Array Operations:
# ● Find the maximum, minimum, and mean temperatures for the week.

temperatures_w1_f = (temperatures_w1 * 9/5) + 32
print(temperatures_w1_f)

print(np.max(temperatures_w1_f))
print(np.min(temperatures_w1_f))
print(np.mean(temperatures_w1_f))
```

```python
# 4. Array Slicing and Indexing:
print(temperatures_w1[0:3])
print(temperatures_w1[-2:])
print(temperatures_w1[2:5])
```

```python
# 5. Create a 2D Array:

temperatures=np.array([[22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],[19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]])
print(temperatures)
```

```python
# 6.  A .Inspect and Slice the 2D Array: 

print (temperatures.shape)
print(temperatures.size)
print(temperatures.dtype)

```

```python
# B. Extract the temperatures for each week and the weekends (last two days) for both weeks.

week_1 = temperatures[0, :]
week_2 = temperatures[1, :]
weekends = temperatures[:, -2:]

print(week_1)
print(week_2)
print(weekends)
```

Pandas Series:  

```python
import pandas as pd
```

```python
# 1. Creating Pandas Series:

marks= pd.Series([95, 92, 89, 85, 80 ],index = ['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])
print (marks)
```

```python
# 2. Indexing and Slicing:

print(marks.iloc[0])

print(marks.loc["Rank1":"Rank3"])

print(marks.loc["Rank3"])

highrank= marks[marks > 90]

highrank
```

```python
# 3. Manipulating Series:
marks.loc['Rank1'] = 100
print(marks)

marks = marks.drop('Rank5')
print(marks)

cgpa = marks / 10
print(cgpa)
```

Pandas DataFrame

```python
# 1. Creating Pandas DataFrame: 

data = {
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture',
                        'Clothing', 'Electronics', 'Furniture', 'Clothing',
                        'Furniture', 'Electronics'],
    'Region': ['North', 'South', 'North', 'East', 'West',
               'North', 'East', 'West', 'South', 'North'],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
}

transactions = pd.DataFrame(data)
transactions
```

3. Manipulating the DataFrame:

```python
# ● Modify the 'Amount' for the transaction with TransactionID 102 to 165. 
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165
transactions
```

```python
# ● Add a new column 'Discount' by calculating 10% of the 'Amount'.

transactions['Discount'] = transactions['Amount'] * 0.10
transactions
```

```python
# ● Remove the row with TransactionID 109. 

transactions = transactions.drop(transactions[transactions['TransactionID'] == 109].index)
transactions
```

```python
# ● Delete the 'Discount' column from the DataFrame.

transactions = transactions.drop(columns=['Discount'])
transactions
```
